# WHILE

```python
soma = 0
contador = 1

while contador <=5:
    numero = int(input("informe um numero: "))
    soma = soma + numero
    contador = contador +1

print(f"Soma: {soma}")
```

# FOR


```python
soma = 0
for nota in [1, 2, 3, 4]:
    valor = float(input(f"Informe a nota {nota}: "))
    soma += valor  

media = soma/4
print(f"Média: {media}") 
```
